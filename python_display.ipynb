```python
# python numbers
import pandas as pd


import matplotlib.pyplot as plt
import numpy as np
import csv
import json
```

```python

#df = name of liked songs file
df = pd.read_csv('Liked_Songs.csv')

```

```python
#show top genres
genres = df['Genres'].dropna(). str.split(',').explode().str.strip()

genre_counts = genres.value_counts()

top15 = genre_counts.head(15)

top15.plot(kind="bar")
plt.title('Top 15 Genres')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python

top15 = df.nlargest(15, 'Danceability')
top15['Label'] = top15['Track Name']+' ('+ top15['Danceability'].astype(str)+ ' )'

top15.plot(kind="bar", x='Label', y='Danceability', legend=False)
plt.title('Top 15 Most Danceable Songs')
plt.xlabel('Song')
plt.ylabel('Danceability')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
print(list(df.columns))
```

```python
#show top genres
explicit = df['Explicit']

explicit_counts = explicit.value_counts().reset_index()
explicit_counts.columns = ['Explicit', 'Count']

explicit_counts['Label'] = explicit_counts['Explicit'].astype(str) + ' (' + explicit_counts['Count'].astype(str) + ')'
explicit_counts.plot(kind="bar", x='Label', y='Count')
plt.title('How Many Songs are Explicit vs Not Explicit')
plt.xlabel('Explicit?')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
